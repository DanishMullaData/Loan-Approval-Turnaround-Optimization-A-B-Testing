```python
import pandas as pd

df = pd.read_excel('ABC_Bank_Loan_AB_Test_Data.xlsx')
print(df.shape)
print(df.head())
```

```python
from scipy import stats

# 1. T-TEST on Turnaround Time (drop NaN first, since dropped-off apps have no turnaround time)
pilot_time = df[df['Group'] == 'Pilot']['Turnaround_Time'].dropna()
bau_time = df[df['Group'] == 'BAU']['Turnaround_Time'].dropna()

t_stat, p_value_ttest = stats.ttest_ind(pilot_time, bau_time)

print("=== T-TEST: Turnaround Time ===")
print(f"Pilot mean: {pilot_time.mean():.2f} days")
print(f"BAU mean: {bau_time.mean():.2f} days")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value_ttest:.10f}")
print(f"Significant? {'YES' if p_value_ttest < 0.05 else 'NO'}")
print()

# 2. CHI-SQUARE TEST on Drop-off Rate
dropoff_table = pd.crosstab(df['Group'], df['Status'] == 'Dropped-off')
chi2_dropoff, p_value_dropoff, dof, expected = stats.chi2_contingency(dropoff_table)

print("=== CHI-SQUARE TEST: Drop-off Rate ===")
print(dropoff_table)
print(f"Chi-square statistic: {chi2_dropoff:.4f}")
print(f"P-value: {p_value_dropoff:.10f}")
print(f"Significant? {'YES' if p_value_dropoff < 0.05 else 'NO'}")
print()

# 3. CHI-SQUARE TEST on Error Rate
error_table = pd.crosstab(df['Group'], df['Error_Flag'])
chi2_error, p_value_error, dof2, expected2 = stats.chi2_contingency(error_table)

print("=== CHI-SQUARE TEST: Error Rate ===")
print(error_table)
print(f"Chi-square statistic: {chi2_error:.4f}")
print(f"P-value: {p_value_error:.10f}")
print(f"Significant? {'YES' if p_value_error < 0.05 else 'NO'}")
```
